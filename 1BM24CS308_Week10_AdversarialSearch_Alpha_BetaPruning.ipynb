```python

tree = [
    [
        [10, 9],
        [14, 18]
    ],
    [
        [5, 4],
        [50, 3]
    ]
]

pruned = []

def collect(node, path):
    if not isinstance(node, list):
        pruned.append((path, node))
        return

    for i, child in enumerate(node):
        side = "Left" if i == 0 else "Right"
        collect(child, path + [side])

def alpha_beta(node, maximizing, alpha, beta, path):
    if not isinstance(node, list):
        return node, path + [str(node)]

    if maximizing:
        value = float("-inf")
        best_path = []

        for i, child in enumerate(node):
            side = "Left" if i == 0 else "Right"

            result, result_path = alpha_beta(
                child, False, alpha, beta, path + [side]
            )

            if result > value:
                value = result
                best_path = result_path

            alpha = max(alpha, value)

            if alpha >= beta:
                for remaining in node[i + 1:]:
                    j = node.index(remaining, i + 1)
                    next_side = "Left" if j == 0 else "Right"
                    collect(remaining, path + [next_side])
                break

        return value, best_path

    else:
        value = float("inf")
        best_path = []

        for i, child in enumerate(node):
            side = "Left" if i == 0 else "Right"

            result, result_path = alpha_beta(
                child, True, alpha, beta, path + [side]
            )

            if result < value:
                value = result
                best_path = result_path

            beta = min(beta, value)

            if beta <= alpha:
                for j in range(i + 1, len(node)):
                    remaining = node[j]
                    next_side = "Left" if j == 0 else "Right"
                    collect(remaining, path + [next_side])
                break

        return value, best_path


value, best_path = alpha_beta(
    tree, True, float("-inf"), float("inf"), ["Root"]
)

print("Root Node Value:", value)
print("Optimal Path:", " -> ".join(best_path))
print("Pruned Paths:")

for path, leaf in pruned:
    print(" -> ".join(path + [str(leaf)]))
```
